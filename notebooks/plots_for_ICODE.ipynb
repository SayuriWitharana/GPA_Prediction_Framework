# Plots for ICODE Submission

This notebook reproduces the RMSE, prediction-stability (SD of RMSE), and bias trend
plots using the exact numbers reported in `conferences/Final_Submission_Extended_Abstract.docx`
(Table I: "Cross-validation results (2017-2018)"), **not** the numbers currently in
`resultPlots.ipynb` (that notebook was corrected after this paper was submitted).

**Data provenance for every point plotted below:**

- Where the ICODE Table I reports a value for a given group/semester, that exact value
  (including sign) is used here.
- Table I does not report every semester for every group (it only shows the semesters
  around each group's convergence point). For those missing semesters, values are taken
  from the current `resultPlots.ipynb` pipeline output, which independently matches the
  paper's numbers wherever the two overlap.
- **Semester 6 is not reported in the paper for any group.** Per instruction, the Semester 6
  point on each line is a carried-forward duplicate of that group's Semester 5 value (not an
  independently measured Semester 6 result). This is flagged explicitly here and in the plot
  styling (dashed segment) so it isn't mistaken for a real measurement.


```python
import matplotlib.pyplot as plt
import pandas as pd

semesters = [0, 1, 2, 3, 4, 5, 6]  # 0 = Pre-semester, 1..6 = Sem1..Sem6

# Values sourced from ICODE Final_Submission_Extended_Abstract.docx, Table I,
# supplemented (only where Table I has no entry) from resultPlots.ipynb's
# current pipeline output. Sem6 = Sem5 carried forward (not in the paper).

data = {
    'Sem': semesters,
    # --- RMSE means ---
    'U_RMSE': [0.498, 0.354, 0.291, 0.274, 0.208, 0.167, 0.146],  # Sem4,Sem5 from paper; Sem6=Sem5
    'A_RMSE': [0.191, 0.195, 0.162, 0.135, 0.109, 0.092, 0.063],  # Sem2,Sem3 from paper; Sem6=Sem5
    'P_RMSE': [0.299, 0.206, 0.159, 0.138, 0.097, 0.079, 0.056],  # Pre-Sem3 from paper; Sem6=Sem5

    # --- RMSE SD (stability) ---
    'U_SD':   [0.117, 0.100, 0.067, 0.081, 0.084, 0.080, 0.069],  # Sem4,Sem5 from paper; Sem6=Sem5
    'A_SD':   [0.055, 0.042, 0.030, 0.030, 0.032, 0.023, 0.017],  # Sem2,Sem3 from paper; Sem6=Sem5
    'P_SD':   [0.051, 0.033, 0.028, 0.024, 0.019, 0.020, 0.009],  # Pre-Sem3 from paper; Sem6=Sem5

    # --- Bias means ---
    'U_Bias': [-0.343, -0.167, -0.088, -0.069, -0.052, -0.035, -0.027],   # Sem4,Sem5 from paper; Sem6=Sem5
    'A_Bias': [-0.007, 0.022, 0.106, 0.025, 0.012, 0.017, 0.004],      # Sem2,Sem3 from paper; Sem6=Sem5
    'P_Bias': [0.182, 0.078, 0.042, -0.028, 0.019, 0.010, 0.007],      # Pre-Sem3 from paper; Sem6=Sem5
}

df_results = pd.DataFrame(data)
df_results
```

```python
# PLOT 1: Group-wise RMSE vs Semester
plt.figure(figsize=(10, 6))

for col, label, color in [('U_RMSE', 'Underperforming', 'red'),
                           ('A_RMSE', 'Average', 'orange'),
                           ('P_RMSE', 'Performing', 'green')]:
    # solid line through Sem0-Sem5 (paper/pipeline-sourced), dashed segment into Sem6 (carried forward)
    plt.plot(df_results['Sem'][:6], df_results[col][:6], marker='o', label=label, color=color)
    plt.plot(df_results['Sem'][5:], df_results[col][5:], linestyle='--', color=color, alpha=0.6)

plt.title("Group-wise RMSE vs Semester (ICODE submission values)", fontsize=14)
plt.xlabel("Semester", fontsize=12)
plt.ylabel("Mean RMSE", fontsize=12)
plt.xticks(semesters)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.savefig('ICODE_RMSE_Trend.png', dpi=300)
plt.show()
```

```python
# PLOT 2: Group-wise Prediction Stability (SD of RMSE) vs Semester
plt.figure(figsize=(10, 6))

for col, label, color in [('U_SD', 'Underperforming', 'red'),
                           ('A_SD', 'Average', 'orange'),
                           ('P_SD', 'Performing', 'green')]:
    plt.plot(df_results['Sem'][:6], df_results[col][:6], marker='o', label=label, color=color)
    plt.plot(df_results['Sem'][5:], df_results[col][5:], linestyle='--', color=color, alpha=0.6)

plt.title("Group-wise Prediction Stability (SD of RMSE) vs Semester (ICODE submission values)", fontsize=14)
plt.xlabel("Semester", fontsize=12)
plt.ylabel("SD of RMSE", fontsize=12)
plt.xticks(semesters)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.savefig('ICODE_Stability_Trend.png', dpi=300)
plt.show()
```

```python
# PLOT 3: Group-wise Mean Bias vs Semester
plt.figure(figsize=(10, 6))

for col, label, color in [('U_Bias', 'Underperforming', 'red'),
                           ('A_Bias', 'Average', 'orange'),
                           ('P_Bias', 'Performing', 'green')]:
    plt.plot(df_results['Sem'][:6], df_results[col][:6], marker='^', label=label, color=color)
    plt.plot(df_results['Sem'][5:], df_results[col][5:], linestyle='--', color=color, alpha=0.6)

plt.axhline(0, color='black', linewidth=1.5, linestyle='-')
plt.title("Group-wise Mean Bias vs Semester (ICODE submission values)", fontsize=14)
plt.xlabel("Semester", fontsize=12)
plt.ylabel("Mean Bias (Actual - Predicted GPA)", fontsize=12)
plt.xticks(semesters)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.savefig('ICODE_Bias_Trend.png', dpi=300)
plt.show()
```

**Note on the dashed Sem5-Sem6 segment:** every chart above breaks the line style between
Semester 5 and Semester 6 to visually mark that the Semester 6 point is not an independent
result — it repeats the Semester 5 value from the paper. Keep this in mind if these figures
are reused anywhere the distinction matters.


## Reviewer Figure: Convergence of the Three Criteria

ICODE reviewer feedback: *"Consider adding a brief visual representation (e.g., a plot) of
the convergence of the three criteria over semesters to improve readability."*

RMSE and Bias don't "converge" in the notebook's sense above — they simply cross a fixed
tolerance threshold (RMSE ≤ 0.25 for High-performing/Average, ≤ 0.35 for Underperforming;
|Bias| ≤ 0.10 for all groups, per the paper's Evaluation Framework). SD is the one metric
with an actual convergent trend. Two candidate figures for the one plot you have space for:

- **Option 1 — 3-panel convergence figure:** the RMSE and Bias plots from above, each with
  their tolerance threshold(s) drawn in, stacked with the SD trend as a third panel. Reads
  as a single figure with panels (a)/(b)/(c).
- **Option 2 — pass/fail grid:** a compact green/red grid showing, per group and semester,
  whether RMSE and Bias are within tolerance. SD is deliberately left out of this grid —
  Table I doesn't define a hard numeric cutoff for when SD counts as "converged," so forcing
  it into a pass/fail cell would mean inventing a threshold that isn't in the paper.


```python
# OPTION 1: 3-panel convergence figure (RMSE with thresholds / Bias with threshold / SD trend)
fig, axes = plt.subplots(3, 1, figsize=(8, 11), sharex=True)

group_style = [('U_RMSE', 'U_SD', 'U_Bias', 'Underperforming', 'red'),
               ('A_RMSE', 'A_SD', 'A_Bias', 'Average', 'orange'),
               ('P_RMSE', 'P_SD', 'P_Bias', 'Performing', 'green')]

# (a) RMSE with accuracy thresholds
ax = axes[0]
for rmse_col, _, _, label, color in group_style:
    ax.plot(df_results['Sem'][:6], df_results[rmse_col][:6], marker='o', label=label, color=color)
    ax.plot(df_results['Sem'][5:], df_results[rmse_col][5:], linestyle='--', color=color, alpha=0.6)
ax.axhline(0.25, color='gray', linestyle='-', linewidth=1.2, label='Tolerance: High-perf & Average (0.25)')
ax.axhline(0.35, color='gray', linestyle='--', linewidth=1.2, label='Tolerance: Underperforming (0.35)')
ax.set_ylabel("Mean RMSE")
ax.set_title("(a) RMSE vs Semester", loc='left', fontsize=11)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(fontsize=8)

# (b) Bias with equity threshold
ax = axes[1]
for _, _, bias_col, label, color in group_style:
    ax.plot(df_results['Sem'][:6], df_results[bias_col][:6], marker='^', label=label, color=color)
    ax.plot(df_results['Sem'][5:], df_results[bias_col][5:], linestyle='--', color=color, alpha=0.6)
ax.axhline(0, color='black', linewidth=1.0)
ax.axhline(0.10, color='gray', linestyle='--', linewidth=1.2, label='Equity threshold (±0.10)')
ax.axhline(-0.10, color='gray', linestyle='--', linewidth=1.2)
ax.set_ylabel("Mean Bias")
ax.set_title("(b) Bias vs Semester", loc='left', fontsize=11)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(fontsize=8)

# (c) SD of RMSE (stability trend, no hard threshold)
ax = axes[2]
for _, sd_col, _, label, color in group_style:
    ax.plot(df_results['Sem'][:6], df_results[sd_col][:6], marker='s', label=label, color=color)
    ax.plot(df_results['Sem'][5:], df_results[sd_col][5:], linestyle='--', color=color, alpha=0.6)
ax.set_ylabel("SD of RMSE")
ax.set_xlabel("Semester")
ax.set_title("(c) Prediction Stability (SD of RMSE) vs Semester", loc='left', fontsize=11)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(fontsize=8)

plt.xticks(semesters)
plt.tight_layout()
plt.savefig('ICODE_Convergence_3Panel.png', dpi=300)
plt.show()
```

```python
# OPTION 2: pass/fail grid for RMSE and Bias (SD excluded - no hard threshold defined in the paper)
import numpy as np
from matplotlib.colors import ListedColormap

rmse_thresholds = {'U_RMSE': 0.35, 'A_RMSE': 0.25, 'P_RMSE': 0.25}
bias_threshold = 0.10

rows = []
row_labels = []
for group_key, group_label in [('U', 'Underperforming'), ('A', 'Average'), ('P', 'Performing')]:
    rmse_col = f'{group_key}_RMSE'
    bias_col = f'{group_key}_Bias'
    rows.append((df_results[rmse_col] <= rmse_thresholds[rmse_col]).astype(int).values)
    row_labels.append(f'{group_label} – RMSE')
    rows.append((df_results[bias_col].abs() <= bias_threshold).astype(int).values)
    row_labels.append(f'{group_label} – Bias')

grid = np.array(rows)

fig, ax = plt.subplots(figsize=(8, 4.5))
cmap = ListedColormap(['#d62728', '#2ca02c'])  # red = fail, green = pass
ax.imshow(grid, cmap=cmap, aspect='auto', vmin=0, vmax=1)

ax.set_xticks(range(len(semesters)))
ax.set_xticklabels(semesters)
ax.set_yticks(range(len(row_labels)))
ax.set_yticklabels(row_labels)
ax.set_xlabel("Semester")
ax.set_title("Criterion Pass/Fail by Group and Semester (RMSE & Bias)", fontsize=12)

# gridlines between cells
ax.set_xticks(np.arange(-0.5, len(semesters), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(row_labels), 1), minor=True)
ax.grid(which='minor', color='white', linewidth=2)
ax.tick_params(which='minor', bottom=False, left=False)

# mark the carried-forward Sem6 column
ax.axvline(5.5, color='white', linewidth=2, linestyle=':')

plt.tight_layout()
plt.savefig('ICODE_Convergence_Grid.png', dpi=300)
plt.show()
```
